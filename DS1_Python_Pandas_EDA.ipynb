```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 20)

print("All libraries imported successfully!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python



from google.colab import files

uploaded = files.upload()

df = pd.read_csv('netflix_titles.csv')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python



print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())

print("\nRandom 5 Records:")
display(df.sample(5, random_state=42))

print("\nDataset Information:")
df.info()

print("\nMemory Usage:")
print(df.memory_usage(deep=True).sum() / (1024**2), "MB")
```

```python



print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

display(missing_percentage.round(2))

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python


missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing.values,
    y=missing.index
)

plt.title("Missing Values in Netflix Dataset")
plt.xlabel("Number of Missing Values")
plt.ylabel("Columns")

plt.tight_layout()
plt.show()
```

```python



eda = df.copy()


eda = eda.drop_duplicates()


for col in eda.select_dtypes(include='object').columns:
    eda[col] = eda[col].str.strip()

eda['date_added'] = pd.to_datetime(
    eda['date_added'],
    errors='coerce'
)


eda['duration_value'] = pd.to_numeric(
    eda['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

print("Data preprocessing completed!")

print("Original Dataset Shape:", df.shape)
print("Processed Dataset Shape:", eda.shape)

display(eda.head())
```

```python



print("Numerical Summary Statistics:")

display(eda.describe())

print("\nComplete Dataset Summary:")

display(eda.describe(include='all').T)

print("\nRelease Year Statistics:")

print("Mean:", eda['release_year'].mean())
print("Median:", eda['release_year'].median())
print("Minimum:", eda['release_year'].min())
print("Maximum:", eda['release_year'].max())
```

```python


print("Movies vs TV Shows:")
display(eda['type'].value_counts())

print("\nContent Ratings:")
display(eda['rating'].value_counts().head(10))

print("\nTop 10 Countries:")
display(
    eda['country']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

print("\nTop 10 Genres:")
display(
    eda['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)
```

```python


plt.figure(figsize=(8, 5))

sns.countplot(
    data=eda,
    x='type'
)

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()
```

```python


year_counts = eda['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker='o'
)

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()
```

```python


movie_duration = eda[
    eda['type'] == 'Movie'
]['duration_value'].dropna()

plt.figure(figsize=(10, 5))

sns.histplot(
    movie_duration,
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

```python



country_counts = (
    eda['country']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=country_counts.values,
    y=country_counts.index
)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python


rating_counts = eda['rating'].dropna().value_counts().head(12)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=rating_counts.values,
    y=rating_counts.index
)

plt.title("Netflix Content Rating Distribution")
plt.xlabel("Number of Titles")
plt.ylabel("Content Rating")

plt.tight_layout()
plt.show()
```

```python



print("========== KEY FINDINGS ==========")


type_counts = eda['type'].value_counts()

print("\n1. Most Common Content Type:")
print(type_counts.index[0], "-", type_counts.iloc[0], "titles")


year_counts = eda['release_year'].value_counts()

print("\n2. Most Common Release Year:")
print(year_counts.index[0], "-", year_counts.iloc[0], "titles")

print("\n3. Average Movie Duration:")

print(
    round(movie_duration.mean(), 2),
    "minutes"
)


print("\n4. Leading Content-Producing Country:")

print(country_counts.index[0])


print("\n5. Most Common Content Rating:")

print(rating_counts.index[0])

print("\n==================================")
```
